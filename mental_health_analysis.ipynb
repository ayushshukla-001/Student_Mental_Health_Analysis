## Introduction

The importance of mental health among college students cannot be emphasized enough. As students leave behind familiar environments and enter university life, they may experience significant emotional, social, and academic challenges. Students from different courses, age groups, and academic years may experience mental health difficulties at some point in their lives.

This project focuses on understanding common mental health concerns among university students, particularly **anxiety, depression, and panic attacks**.

### Anxiety

Anxiety is an emotional state characterized by feelings of nervousness, worry, restlessness, or tension. It may also involve excessive concern about future events and can be accompanied by physical and behavioral symptoms.

Common symptoms of anxiety include:

* Disproportionate feelings of nervousness, restlessness, or tension
* A sense of impending danger, panic, or doom
* Rapid breathing or hyperventilation
* Trembling or sweating
* Weakness and fatigue
* Difficulty falling asleep or staying asleep
* Changes in appetite, including eating too little or too much
* Nausea or headaches

### Depression

Depression is a mental health condition that can affect a person's thoughts, emotions, behavior, motivation, and overall sense of well-being.

Common symptoms of depression include:

* Persistent feelings of sadness or hopelessness
* Changes in mood
* Changes in sleep or appetite
* Withdrawal from social activities and increased isolation
* Increased pessimism or negative thinking
* Low energy or feeling lethargic
* Difficulty concentrating or completing tasks
* Loss of interest or enjoyment in previously enjoyable activities

### Panic Attack

A panic attack is a sudden episode of intense fear or discomfort. It may occur unexpectedly and can produce both psychological and physical symptoms.

Common symptoms of a panic attack include:

* Rapid heartbeat or palpitations
* Sweating
* Chest pain or discomfort
* Shortness of breath
* Trembling or shaking
* Dizziness
* Numbness or tingling
* Confusion
* A feeling of losing control or impending danger

Symptoms often reach their peak within a short period, although the duration and intensity can vary from person to person.

##### Few findings

- Year 1  and year 1 mean the same thing (and same with other values) yet are interpreted as different.

```python
def Clean(Text):
    Text = Text[-1]
    Text = int(Text)
    return Text
data["Year"] = data["Year"].apply(Clean)
print("First three values of Year after cleaning text:")
print(data["Year"][:3], "\n")
```

##### Few findings

```python
def remove_space(string):
    string = string.strip()
    return string
data["CGPA"] = data["CGPA"].apply(remove_space)
print("First three values of CGPA after cleaning text:")
print(data["CGPA"][:3], "\n")
print(data['CGPA'].unique())
```

## Model Selection

For model selection, I will be building pipelines of five different classifiers and select one with the best fit results.

In this section:
- Split data into trainig and testing sets
- Assigning targets and features
- The model pipelines with preprocessing:
- Fitting the training set to the various models.
- Getting the confusion matrix and accuracy scores.
- Picking the best classifier.

```python
X = data.drop(["CGPA"],axis=1)
y = data["CGPA"]
```

```python
#spliting test and training sets
X_train, X_test, y_train,y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

```python
X_train['Gender'] = X_train['Gender'].map({'Male': 0, 'Female': 1})
X_test['Gender'] = X_test['Gender'].map({'Male': 0, 'Female': 1})

# Convert categorical columns
X_train = pd.get_dummies(X_train, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

# Make columns same
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)

# Remove missing values
X_train = X_train.fillna(0)
X_test = X_test.fillna(0)


#A quick model selection process
pipeline_lr=Pipeline([('lr_classifier',LogisticRegression(random_state=42))])

pipeline_dt=Pipeline([('dt_classifier',DecisionTreeClassifier(random_state=42))])

pipeline_rf=Pipeline([('rf_classifier',RandomForestClassifier())])

pipeline_svc=Pipeline([('sv_classifier',SVC())])

pipelines = [pipeline_lr, pipeline_dt, pipeline_rf, pipeline_svc]

pipe_dict = {
    0: 'Logistic Regression',
    1: 'Decision Tree',
    2: 'RandomForest',
    3: 'SVC'
}

# Fit the pipelines
for pipe in pipelines:
    pipe.fit(X_train, y_train)

# Cross validation on accuracy
cv_results_accuracy = []

for i, model in enumerate(pipelines):
    cv_score = cross_val_score(model, X_train, y_train, cv=10)
    cv_results_accuracy.append(cv_score)
    print("%s: %f " % (pipe_dict[i], cv_score.mean()))
```

So Random Forest does best amongst the models to be the most accurate. Let us build a better random forest with grid search cv. Let's find out how it performs on testset
